```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/selva86/datasets/master/Cars93_miss.csv"
df = pd.read_csv(url)
```

```python
print(" Data shape:", df.shape)
df['Price'].fillna(df['Price'].median(), inplace=True)
df['MPG.city'].fillna(df['MPG.city'].mean(), inplace=True)


plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Horsepower', y='Price', hue='Type', palette='Set2')
plt.title('Relationship between Horsepower and Price')
plt.xlabel('Horsepower')
plt.ylabel('Price ($1000)')
plt.savefig('horsepower_vs_price.png')
plt.show()
```

```python
plt.figure(figsize=(8,6))
numeric_df =df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(),annot=False, cmap='Blues')
plt.title('Feature Correlation Matrix')
plt.savefig('correlation_matrix.png')
plt.show()
```
